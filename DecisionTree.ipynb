```python
#using decision tree regression model to predict the salary of new recriut based on his or her previous company data
#previous salaray is said to be 160000 based on position and level
#level 6.5
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Position_Salaries.csv')
x = dataset.iloc[ : , 1 :  -1].values
y = dataset.iloc[ : , -1].values

print(x)
print(y)
```

```python
#Training the model using Decision tree model
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor.fit(x, y)
```

```python
regressor.predict([[6.5]])
```

```python
#visualizing the results in higher resolution
#2d dataset are not good visually in decision tree model but either ways here is the visual
x_grid = np.arange(min(x), max(x), 0.01)
x_grid = x_grid.reshape((len(x_grid), 1))
plt.scatter(x, y, color = 'red')
plt.plot(x_grid, regressor.predict(x_grid), color = 'blue')
plt.title('Salary of new recriut (Decision Tree)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```
